Every sub-halo of step halo_redshift_int gets the total spectrum of age age age_myr[k-1]. Each halo has its own metallicity $Z_h$ that is between two G19 metallicities from the G19 grid. $Z_{low} \leq Z_h < Z_{high}$. So $Z_h < 0.0002$ is clipped to 0.0002 with a weight defined by:

$$w_Z = \frac{logZ_h - log Z_{low}}{log Z_{high}-log Z_{low}}$$

$L_{\lambda, delphi}^h = \frac{M_h^{\rm new}}{10^6\,M_\odot}\left[(1 - w_Z)\,L_{\lambda, G19}(Z_{\rm low}, {\rm age}_k) + w_Z\,L_{\lambda, G19}(Z_{\rm high}, {\rm age}_k)\right]$, with $M_h^{\rm new}$ = SFR $\times$ 30 Myr.

The spectrum table spectra[a, i, :]

Everything below is a linear combination of this one table, so it is the only large array that is computed (44 × 4 × 1000).

* a = step index = halo_redshift_int - 1, age age_myr[a] (2, 30, 60, … Myr): the age at z = 4.5 of stars formed at that step.
* i = metallicity index into Z_grid = [0.0002, 0.002, 0.006, 0.014].
* each row is $L_\lambda$ [erg/s/Å] of a $10^6\,M_\odot$ burst = Starburst99 single stars + the G19 stripped-star row nearest in time (no interpolation over the 500 G19 ages).

# Metallicity Weights

For each halo's $Z_h$, we need an index of $i_{low}$ of $Z_{low}$ and $w$ which is the weight of the grid values just above. 

$L_{\lambda, G19}(Z_h) = (1 - w)\,L_{\lambda, G19}(Z_{low}) + w\,L_{\lambda, G19}(Z_{high}), \qquad w = \dfrac{\log Z_h - \log Z_{low}}{\log Z_{high} - \log Z_{low}}$


searchsorted counts how many grid values are less than Z. Minus 1 is the index of the largest one i.e. $Z_{low}$. The outer clip to [0, 2] which is subtantiala t Z = 0.014. It gives the pair (0.0006, 0.014) with w=1, which is the same spectrum. 


| $Z_h$ | i_low | ($Z_{low}$, $Z_{high}$) | w |
|---|---|---|---|
| 0, 0.000056 | 0 | (0.0002, 0.002) | 0 |
| 0.001 | 0 | (0.0002, 0.002) | 0.699 |
| 0.002 | 1 | (0.002, 0.006) | 0 |
| 0.003451 | 1 | (0.002, 0.006) | 0.497 |
| 0.014+ | 2 | (0.006, 0.014) | 1 |

```python
def metallicity_weights(init_metallicity):
    Z = np.clip(init_metallicity, Z_grid[0], Z_grid[-1])                                   #keep Z inside [0.0002, 0.014]
    i_low = np.clip(np.searchsorted(Z_grid, Z, side="right") - 1, 0, len(Z_grid) - 2)     #index of Z_low (0, 1 or 2)
    w = (np.log10(Z) - np.log10(Z_grid[i_low])) / (np.log10(Z_grid[i_low + 1]) - np.log10(Z_grid[i_low]))   #weight of Z_high
    return i_low, w
```

```python
k = halo_table["halo_redshift_int"].to_numpy()
i_low, w_Z = metallicity_weights(halo_table["init_metallicity"].to_numpy())
halo_table["age_myr"] = age_myr[k - 1]
halo_table["Z_low"] = Z_grid[i_low]
halo_table["Z_high"] = Z_grid[i_low + 1]
halo_table["w_Z"] = w_Z
halo_table["new_stars"] = halo_table["sfr"] * 30e6        #in Msun
```

# Halo Spectrum

```python
def halo_spectrum(j):
    L = (1 - w_Z[j]) * spectra[k[j] - 1, i_low[j]] + w_Z[j] * spectra[k[j] - 1, i_low[j] + 1]
    return halo_table.at[j, "new_stars"] / 1e6 * L
```

```python
norm_Z = plt.matplotlib.colors.LogNorm(Z_grid[0], Z_grid[-1])
cmap_Z = gcm.pan()
sfh_spectrum = np.zeros((len(age_myr), wavelength.size))              #summed spectrum of every step (star formation history), step x wavelength
for step_plot in range(1, len(age_myr) + 1):
    halos_plot = halo_table.index[(halo_table["halo_redshift_int"] == step_plot) & (halo_table["sfr"] > 0)]
    if len(halos_plot) == 0:
        continue
    Z_plot = halo_table.loc[halos_plot, "init_metallicity"].clip(Z_grid[0], Z_grid[-1]).to_numpy()   #clipped, as in metallicity_weights
    fig = plt.figure(figsize=(8, 6))
    L_all = np.array([halo_spectrum(j) for j in halos_plot])            #halos x wavelength
    sfh_spectrum[step_plot - 1] = L_all.sum(axis=0)                     #sum over all halos of the step
    color = cmap_Z(norm_Z(Z_plot))
    for i in range(L_all.shape[0]):
        plt.plot(wavelength, L_all[i, :], linewidth = 0.5, alpha = 1, color = color[i])
    plt.plot(wavelength, sfh_spectrum[step_plot - 1], color="k", linewidth=2, label="sum of the step")
    plt.gca().set(xscale='log', yscale = 'log')
    plt.gca().set_xlim(100, 1e4) 
    plt.gca().set_title(f"step {step_plot}: {len(halos_plot)} halos at {age_myr[step_plot - 1]} Myr")
    plt.legend(frameon=False)
    plt.show()
```

```python
total_spectrum = sfh_spectrum.sum(axis=0)                            #sum of all steps: the spectrum of the tree at z = 4.5
norm_age = plt.Normalize(age_myr.min(), age_myr.max())
plt.figure(figsize=(10, 6))
for i_step in range(len(age_myr)):
    if sfh_spectrum[i_step].any():
        plt.plot(wavelength, sfh_spectrum[i_step], color=gcm.pan()(norm_age(age_myr[i_step])), linewidth=0.8)
plt.plot(wavelength, total_spectrum, color="k", linewidth=2, label="sum of all steps")
plt.gca().set(xscale='log', yscale='log')
plt.gca().set_xlim(100, 1e4)
plt.gca().set_ylim(total_spectrum.max() * 1e-8, total_spectrum.max() * 3)
plt.gca().set_xlabel(r"$\lambda$ [$\AA$]")
plt.gca().set_ylabel(r"$L_\lambda$ [erg s$^{-1}$ $\AA^{-1}$]")
plt.gca().set_title(f"tree {index_tree}: summed spectrum of every step")
plt.colorbar(plt.cm.ScalarMappable(norm=norm_age, cmap=gcm.pan()), ax=plt.gca(), label="age of the step [Myr]")
plt.legend(frameon=False)
plt.show()
```
